```python
from google.colab import files
from pathlib import Path

input_dir = Path('../dataset')
input_dir.mkdir(exist_ok=True)
for name, content in files.upload().items():
    (input_dir / name).write_bytes(content)
print('Upload one .xlsx workbook, term_manifest.csv, and scope_mapping.csv.')
```

```python
# Replace the workbook name with the uploaded, approved source.
from subprocess import run
from shutil import make_archive

WORKBOOK = '../dataset/20260926 Offered Courses .xlsx'
output_dir = Path('scripts/colab_prototype/output')
run([
    'python', 'scripts/colab_prototype/preprocess_historical_dataset.py',
    '--workbook', WORKBOOK, '--term-manifest', '../dataset/term_manifest.csv',
    '--scope-mapping', '../dataset/scope_mapping.csv',
    '--title-crosswalk', 'scripts/colab_prototype/historical_title_crosswalk.csv',
    '--code-crosswalk', 'scripts/colab_prototype/historical_code_crosswalk.csv',
    '--excluded-course-codes', 'scripts/colab_prototype/historical_excluded_course_codes.csv',
    '--output-dir', str(output_dir),
], check=True)
print((output_dir / 'source_manifest.json').read_text())
files.download(make_archive('/content/chronos-historical-preprocessing', 'zip', output_dir))
```
